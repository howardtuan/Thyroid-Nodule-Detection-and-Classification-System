# 這個會把txt跟png放同一個資料夾 == 我懶得改

```python
import os
import shutil

# 來源標記檔資料夾（正確的 .txt 檔）
source_label_dir = r"C:\Users\howar\OneDrive\桌面\ALL_YOLO標記檔"

# 目標 YOLO 資料夾（包含 images/ 和 label/ 等）
target_base_dir = r"C:\temp\以舊版的分類影像分布補上新的正常影像_作為偵測用"  # <<< 你需要修改這行

# 遞迴尋找所有圖片並補上對應標記檔
count = 0
for root, dirs, files in os.walk(target_base_dir):
    for file in files:
        if file.lower().endswith(('.jpg', '.jpeg', '.png')):
            name, _ = os.path.splitext(file)
            src_txt = os.path.join(source_label_dir, f"{name}.txt")
            dst_txt = os.path.join(root, f"{name}.txt")

            if os.path.exists(src_txt):
                shutil.copyfile(src_txt, dst_txt)
                count += 1

print(f"已補上 {count} 個 YOLO 標記檔到：{target_base_dir}")
```
