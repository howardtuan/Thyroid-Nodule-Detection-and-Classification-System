```python
import os
import pandas as pd
from tqdm import tqdm
from charset_normalizer import from_path

root_dir = r"D:\TWB_IMG"  # 修改成你的路徑
output_file = "result.txt"
error_log = "error_log.txt"
matched_codes = []

# 自動編碼偵測
def detect_encoding(file_path):
    try:
        result = from_path(file_path).best()
        return result.encoding if result else "utf-8"
    except:
        return "utf-8"

# 嘗試讀取 CSV，允許 header=1 與 fallback header=0
def read_csv_flexibly(path, encoding):
    try:
        df = pd.read_csv(path, encoding=encoding, header=1, dtype=str)
        if len(df) == 0:
            # fallback：若 header=1 無資料，改用 header=0
            df = pd.read_csv(path, encoding=encoding, header=0, dtype=str)
        return df
    except Exception:
        # fallback 編碼：Big5
        return pd.read_csv(path, encoding="big5", header=0, dtype=str)

# 主程式
patient_dirs = [d for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, d))]

# 清空錯誤紀錄
with open(error_log, "w", encoding="utf-8") as f:
    f.write("")

for patient in tqdm(patient_dirs, desc="查詢中"):
    patient_path = os.path.join(root_dir, patient)
    csv_files = [f for f in os.listdir(patient_path) if f.lower().endswith(".csv")]
    if not csv_files:
        continue

    csv_path = os.path.join(patient_path, csv_files[0])

    try:
        encoding = detect_encoding(csv_path)
        df = read_csv_flexibly(csv_path, encoding)

        if df.empty:
            raise ValueError("CSV 無有效資料")

        # 取第一列最後一欄
        last_text = str(df.iloc[0, -1]).strip()

        print(f"\n--- 病人代碼: {patient} ---")
        print(f"使用編碼：{encoding}")
        print(f"總結內容：{last_text}")

        if "結節" in last_text and "單一" in last_text :
            print("✅ 符合條件")
            matched_codes.append(patient)
        else:
            print("❌ 不符合條件")

    except Exception as e:
        print(f"\n❌ 讀取錯誤: {csv_path}，錯誤訊息: {e}")
        with open(error_log, "a", encoding="utf-8") as f:
            f.write(f"{patient}: {e}\n")

# 輸出結果
with open(output_file, "w", encoding="utf-8") as f:
    for code in matched_codes:
        f.write(code + "\n")

print(f"\n✅ 查詢完成，找到 {len(matched_codes)} 筆結果")
print(f"📄 符合條件者寫入：{output_file}")
print(f"⚠️ 讀取錯誤者寫入：{error_log}")
```
