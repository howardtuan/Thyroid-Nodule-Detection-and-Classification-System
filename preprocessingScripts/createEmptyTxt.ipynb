# 幫正常影像做yolo標記檔

```python
import os

# 原始影像所在資料夾
image_dir = r"C:\Users\howar\OneDrive\桌面\正常_ALLTWB\oriPic"

# 標記檔要輸出的新資料夾
label_dir =r"C:\Users\howar\OneDrive\桌面\正常_ALLTWB\label"
os.makedirs(label_dir, exist_ok=True)

# 支援的圖片副檔名
image_exts = ('.jpg', '.jpeg', '.png')

# 建立同名空白 .txt 標記檔
created = 0
for filename in os.listdir(image_dir):
    if filename.lower().endswith(image_exts):
        name, _ = os.path.splitext(filename)
        txt_path = os.path.join(label_dir, f"{name}.txt")
        if not os.path.exists(txt_path):
            with open(txt_path, 'w') as f:
                pass  # 建立空白內容
            created += 1

print(f"已建立 {created} 個空白 YOLO 標記檔於：{label_dir}")
```
