# 分類資料夾切分 64:16:20

```python
import os
import shutil
import random

# 修改為你實際的資料夾路徑
benign_dir = r'C:\Users\howar\OneDrive\桌面\良性85+115\[還沒裁切成ROI，twb+public]良性all'          # 良性影像資料夾
malignant_dir = r'C:\Users\howar\OneDrive\桌面\惡性200\[還沒裁切成ROI]惡性all'    # 惡性影像資料夾
output_base = r'C:\temp\[Classification]normal_b_m_All_400'  # 切分後輸出資料夾

# 分割比例
train_ratio = 0.64
val_ratio = 0.16
test_ratio = 0.20

# 建立 train/val/test + 子類別資料夾
for split in ['train', 'val', 'test']:
    for cls in ['benign', 'malignant']:
        os.makedirs(os.path.join(output_base, split, cls), exist_ok=True)

# 收集所有圖檔
benign_imgs = [os.path.join(benign_dir, f) for f in os.listdir(benign_dir)
               if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
malignant_imgs = [os.path.join(malignant_dir, f) for f in os.listdir(malignant_dir)
                  if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

# 加上類別標籤並合併
all_images = [(img, 'benign') for img in benign_imgs] + [(img, 'malignant') for img in malignant_imgs]

# 隨機打亂
random.seed(42)
random.shuffle(all_images)

# 切分
total = len(all_images)
train_end = int(total * train_ratio)
val_end = train_end + int(total * val_ratio)

train_set = all_images[:train_end]
val_set = all_images[train_end:val_end]
test_set = all_images[val_end:]

# 複製影像到對應資料夾
def copy_images(image_list, split):
    for src_path, label in image_list:
        filename = os.path.basename(src_path)
        dst_path = os.path.join(output_base, split, label, filename)
        shutil.copyfile(src_path, dst_path)

copy_images(train_set, 'train')
copy_images(val_set, 'val')
copy_images(test_set, 'test')

# 印出統計
def summarize(dataset, name):
    total = len(dataset)
    b = sum(1 for _, lbl in dataset if lbl == 'benign')
    m = total - b
    print(f"{name}: Total={total}, Benign={b}, Malignant={m}")

summarize(train_set, "Train")
summarize(val_set, "Validation")
summarize(test_set, "Test")
```

```python
import os
import shutil
import random

# 設定原始正常影像資料夾
normal_dir = r'C:\Users\howar\OneDrive\桌面\正常_ALLTWB\oriPic'  # 你的正常影像資料夾
output_base = r'C:\Users\howar\OneDrive\桌面\正常_ALLTWB\切分後的'  # 輸出切分後的資料夾

# 切分比例
train_ratio = 0.64
val_ratio = 0.16
test_ratio = 0.20

# 建立資料夾
for split in ['train', 'val', 'test']:
    os.makedirs(os.path.join(output_base, split), exist_ok=True)

# 收集所有影像檔案
normal_images = [f for f in os.listdir(normal_dir) if f.lower().endswith(('.jpg', '.png', '.jpeg'))]
random.seed(42)
random.shuffle(normal_images)

# 計算切分點
total = len(normal_images)
train_end = int(total * train_ratio)
val_end = train_end + int(total * val_ratio)

train_set = normal_images[:train_end]
val_set = normal_images[train_end:val_end]
test_set = normal_images[val_end:]

# 複製影像到對應資料夾
def copy_images(file_list, split):
    for fname in file_list:
        src = os.path.join(normal_dir, fname)
        dst = os.path.join(output_base, split, fname)
        shutil.copyfile(src, dst)

copy_images(train_set, 'train')
copy_images(val_set, 'val')
copy_images(test_set, 'test')

# 印出統計
print(f"Train: {len(train_set)} images")
print(f"Validation: {len(val_set)} images")
print(f"Test: {len(test_set)} images")
```
